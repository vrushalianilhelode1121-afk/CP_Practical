## 4. Activity 1 – Purity-based Classification

### Problem

Write a Python program that accepts the purity percentage of a chemical product and classifies it as follows:

- Purity ≥ 98 → Grade A
- Purity ≥ 95 → Grade B
- Purity ≥ 90 → Grade C
- Otherwise → Rejected

````python
# Activity 1
# Enter purity and classify the product.
# Write your solution here.
purity = float(input("Enter the purity of the chemical: "))
print(f"The entered purity is: {purity}")
if purity >= 98:
    grade = "Grade A"
elif purity >= 95:
    grade = "Grade B"
elif purity >= 90:
    grade = "Grade C"
else:
    grade = "Rejected"

print(f"The product is classified as: {grade}")
````

## 5. Activity 2 – Add pH Checking

Now extend your program so that pH is also considered.

For a batch to receive **Grade A**, both conditions must be satisfied:

- Purity ≥ 98%
- pH between 6.5 and 7.5

If the purity is excellent but the pH is outside the Grade A range, the product must not be classified as Grade A.

### Think before coding

What logical operator is required when **both conditions must be true**?


### Copilot Activity 2 – Understand `and`

Ask Copilot:

Then modify your program to include pH.


````python
# Activity 2
# Add pH as a quality parameter.
# Write/modify your code here.
purity = float(input("Enter the purity of the chemical: "))
print(f"The entered purity is: {purity}")
pH = float(input("Enter the pH of the product: "))
print(f"The entered pH is: {pH}")

if purity >= 98 and pH >= 6 and pH <= 8:
    grade = "Grade A"
elif purity >= 95 and pH >= 6 and pH <= 8:
    grade = "Grade B"
elif purity >= 90 and pH >= 6 and pH <= 8:
    grade = "Grade C"
else:
    grade = "Rejected"

print(f"The product is classified as: {grade}")
````

## 6. Activity 3 – Add Moisture Content

Now include moisture content.

````python
# Activity 3
# Complete chemical quality grading using:
# purity, pH and moisture.
# Write your solution here.
purity = float(input("Enter the purity of the chemical: "))
print(f"The entered purity is: {purity}")
pH = float(input("Enter the pH of the product: "))
print(f"The entered pH is: {pH}")
Moisture = float(input("Enter the moisture content of the product (as a percentage): "))
print(f"The entered moisture content is: {Moisture}")

if purity >= 98 and pH >= 5 and pH <= 8 and Moisture <= 1:
    grade = "Grade A"
elif purity >= 95 and pH >= 6 and pH <= 7.5 and Moisture <= 2:
    grade = "Grade B"
elif purity >= 90 and pH >= 5 and pH <= 9 and Moisture <= 4:
    grade = "Grade C"
else:
    grade = "Rejected"

print(f"The product is classified as: {grade}")

````

## 7. Activity 4 – Input Validation

A real program should not accept obviously invalid measurements.

For this exercise, use these basic checks:

- Purity must be between 0 and 100.
- pH must be between 0 and 14.
- Moisture must not be negative.

If an invalid value is entered, display an appropriate message instead of performing the quality grading.

### Copilot challenge

Ask Copilot:

````python
purity = float(input( "Enter the purity of the chemical:"))
print(f"The entered purity is: {purity}")
if purity < 0 or purity > 100:
    print("Invalid purity value. Please enter a value between 0 acnd 100.")
    exit(1)
pH = float(input("Enter the pH of the product: "))
print(f"The entered pH is: {pH}")
if pH < 0 or pH > 14:
    print("Invalid pH value. Please enter a value between 0 and 14.")
    exit(1)
Moisture = float(input("Enter the moisture content of the product (as a percentage): "))
print(f"The entered moisture content is: {Moisture}")
if Moisture < 0 or Moisture > 100:
    print("Invalid moisture value. Please enter a value between 0 and 100.")
    exit(1)

if purity >= 98 and pH >= 5 and pH <= 8 and Moisture <= 1:
    grade = "Grade A"
elif purity >= 95 and pH >= 6 and pH <= 7.5 and Moisture <= 2:
    grade = "Grade B"
elif purity >= 90 and pH >= 5 and pH <= 9 and Moisture <= 4:
    grade = "Grade C"
else:
    grade = "Rejected"

print(f"The product is classified as: {grade}")
````

## 12. Final Program

Combine your work into one clean final program.

Your final program should:

2. Validate the input values.
