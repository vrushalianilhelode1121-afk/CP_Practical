```python
# Student grade A -above 95 Marks, B - 94-85 Marks, C - 75-84 Marks, D - 65-74 Marks, F - Below 65 Marks
# if elif else format
marks = 85
if marks > 95:
    grade = "A"
elif marks > 94:
    grade = "B"
elif marks > 74:
    grade = "C"
else:
    grade = "F"

print(f"Your grade is: {grade}")
```
